#Crear la primera tabla (customers)

```python
%sql
CREATE OR REPLACE TABLE olist_proyecto.bronce.customers AS
SELECT *
FROM read_files(
  '/Volumes/olist_proyecto/bronce/raw/olist_customers_dataset.csv',
  format => 'csv',
  header => true,
  inferSchema => false
);
```

#Verificar

```python
%sql
SELECT COUNT(*) AS total_filas FROM olist_proyecto.bronce.customers;
SELECT * FROM olist_proyecto.bronce.customers LIMIT 5;
```

#Cargar las tablas

```python
%sql
CREATE OR REPLACE TABLE olist_proyecto.bronce.orders AS
SELECT * FROM read_files('/Volumes/olist_proyecto/bronce/raw/olist_orders_dataset.csv',
  format => 'csv', header => true, inferSchema => false);

CREATE OR REPLACE TABLE olist_proyecto.bronce.order_items AS
SELECT * FROM read_files('/Volumes/olist_proyecto/bronce/raw/olist_order_items_dataset.csv',
  format => 'csv', header => true, inferSchema => false);

CREATE OR REPLACE TABLE olist_proyecto.bronce.order_payments AS
SELECT * FROM read_files('/Volumes/olist_proyecto/bronce/raw/olist_order_payments_dataset.csv',
  format => 'csv', header => true, inferSchema => false);

CREATE OR REPLACE TABLE olist_proyecto.bronce.products AS
SELECT * FROM read_files('/Volumes/olist_proyecto/bronce/raw/olist_products_dataset.csv',
  format => 'csv', header => true, inferSchema => false);

CREATE OR REPLACE TABLE olist_proyecto.bronce.sellers AS
SELECT * FROM read_files('/Volumes/olist_proyecto/bronce/raw/olist_sellers_dataset.csv',
  format => 'csv', header => true, inferSchema => false);

CREATE OR REPLACE TABLE olist_proyecto.bronce.geolocation AS
SELECT * FROM read_files('/Volumes/olist_proyecto/bronce/raw/olist_geolocation_dataset.csv',
  format => 'csv', header => true, inferSchema => false);

CREATE OR REPLACE TABLE olist_proyecto.bronce.category_translation AS
SELECT * FROM read_files('/Volumes/olist_proyecto/bronce/raw/product_category_name_translation.csv',
  format => 'csv', header => true, inferSchema => false);
```

#Verificar que las 7 tablas se crearon

```python
%sql
SELECT 'orders' AS tabla, COUNT(*) AS filas FROM olist_proyecto.bronce.orders
UNION ALL SELECT 'order_items', COUNT(*) FROM olist_proyecto.bronce.order_items
UNION ALL SELECT 'order_payments', COUNT(*) FROM olist_proyecto.bronce.order_payments
UNION ALL SELECT 'products', COUNT(*) FROM olist_proyecto.bronce.products
UNION ALL SELECT 'sellers', COUNT(*) FROM olist_proyecto.bronce.sellers
UNION ALL SELECT 'geolocation', COUNT(*) FROM olist_proyecto.bronce.geolocation
UNION ALL SELECT 'category_translation', COUNT(*) FROM olist_proyecto.bronce.category_translation;
```
